# EfficientNet-B0

Power of EfficientNet: **Compound Scaling**

$$\text{depth}: d = \alpha^{\phi} \quad \text{width}: w = \beta^{\phi} \quad \text{resolution}: r = \gamma^{\phi}$$

subjects to: 
$$\alpha \cdot \beta^{2} \cdot \gamma^{2} \approx 2$$

With:
- $\phi$ = scaling coefficient, for B0 ($\phi=1$)
- $\alpha, \beta, \gamma$ = constants found via grid search

### MBConv (core building block)
1. Expansion (1x1 conv)
$$x_{exp} = \sigma(W_{exp} \times x)$$
2. Depthwise Convolution
$$x_{dw} = \sigma(W_{dw} \odot x_{exp})$$
3. Squeeze-and-Excitation (SE)
$$s = \sigma(W_{2} \cdot \delta(W_{1} \cdot GAP(x_{dw})))$$
4. Projection (1x1 conv)
$$x_{out} = W_proj \times x_{se}$$

Additional Information:
- $\sigma$: sigmoid activation function
$$\sigma(x) = \frac{1}{1 + e^{-x}}$$
- $\delta$: ReLU activation
$$\delta(x) = \max(0, x)$$
- $\text{Swish / SiLU}$: Swish activation function
$$\text{SiLU}(x) = x\cdot \sigma(x)$$
- GAP: Global Average Pooling
$$GAP(x_{c}) = \frac{1}{H \cdot W} \sum_{i=1}^{H} \sum_{j=1}^{W} x_{c}(i, j)$$

```python
# Activation Function
def linear(x: float) -> float:
    """compute linear activation function"""
    return x

def sigmoid(x: float) -> float:
    """compute sigmoid activation function"""
    return 1 / (1 + (np.exp(-x)))

def relu(x: float) -> float:
    """compute relu activation function"""
    return np.maximum(0, x)

def swish(x: float) -> float:
    """compute swish/silu activation function"""     
    return x * sigmoid(x)
```

```python
# Global Average Pooling for EfficientNet
def global_average_pooling(x: np.ndarray) -> float:
    """global average pooling layer"""
    return np.mean(x, axis=(1, 2))
```

```python
# Compound Scaling for EfficientNet
def compound_scaling(phi: int, alpha: float, beta: float, gamma: float) -> tuple:
    """
    Compound scaling method for EfficientNet that makes a balance between depth, width, and resolution of the network.

    Args:
        phi (int): scaling coefficient
        alpha (float): depth of the input
        beta (float): width of the input
        gamma (float): resolution of the input

    Returns:
        tuple: _description_
    """
    depth = alpha ** phi
    width = beta ** phi
    resolution = gamma ** phi
    
    return depth, width, resolution
```
